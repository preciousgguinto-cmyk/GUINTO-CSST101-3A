3. Connect Table Arguments and Evidence

```r
knowledge_base <- merge(
  arguments,
  evidence,
  by = "Argument_ID"
)
print(knowledge_base)
```

4. Convert Evidence Strength into a Score

```r
get_score <- function(strength){
  if (strength == "Strong"){
    return (3)
  } else if (strength == "Moderate"){
    return (2)
  } else if (strength == "Weak"){
    return (1)
  } else {
    return (0)
  }

}

get_score("Moderate")
```

5. Add Evidence Scores:

```r
knowledge_base$Evidence_Score <- sapply(
  knowledge_base$Strength,
  get_score
)

print(knowledge_base)
```

6. Convert Evidence Score to Support

```r
classify_support <- function(score){
  if (score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else {
    return("No Supported")
  }
}

classify_support(3)
```

7. Add Evidence Support

```r
knowledge_base$Support_Level <- sapply(
  knowledge_base$Evidence_Score,
  classify_support
)

print(knowledge_base)
```

9. Laboraty Challenge

```r
library(ggplot2)

# ==============================================================================
# 1. DATA INITIALIZATION & EXTENSIONS
# ==============================================================================

# Table 1: Debate Arguments (Extended with More Arguments & Additional Topics)
debate_arguments <- data.frame(
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006", "A007", "A008", "A009"),
  Position = c("For", "For", "For", "Against", "Against", "Against", "For", "Against", "For"),
  Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity",
    "Dependency", "Assessment", "Accessibility", "Bias & Equity", "Personalization"
  ),
  Claim = c(
    "AI can provide students with immediate learning support.",
    "AI can help students complete routine academic tasks more efficiently.",
    "Allowing AI use can help students develop AI literacy.",
    "Undisclosed AI-generated work may create academic integrity concerns.",
    "Excessive AI use may reduce students' independent problem-solving practice.",
    "AI use may make it more difficult to determine whether submitted work reflects student learning.",
    "AI tools offer customized assistive learning capabilities for neurodivergent students.",
    "Data biases in AI models can lead to unfair assessment outcomes for marginalized groups.",
    "AI adaptively adjusts learning material pace based on individual student progress."
  ),
  stringsAsFactors = FALSE
)

# Table 2: Evidence (Extended with More Evidence Records)
evidence <- data.frame(
  Evidence_ID = c("E001", "E002", "E003", "E004", "E005", "E006", "E007", "E008", "E009"),
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006", "A007", "A008", "A009"),
  Evidence_Type = c("Research", "Research", "Policy", "Policy", "Research", "Research", "Research", "Policy", "Research"),
  Strength = c("Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong", "Strong", "Moderate", "Strong"),
  stringsAsFactors = FALSE
)

# Combine datasets into a master system table
debate_system <- merge(debate_arguments, evidence, by = "Argument_ID")


# ==============================================================================
# 2. MODULAR FUNCTIONS
# ==============================================================================

# Function: Add a new argument and corresponding evidence dynamically
add_record <- function(system_df, arg_id, position, topic, claim, ev_id, ev_type, strength) {
  new_entry <- data.frame(
    Argument_ID = arg_id,
    Position = position,
    Topic = topic,
    Claim = claim,
    Evidence_ID = ev_id,
    Evidence_Type = ev_type,
    Strength = strength,
    stringsAsFactors = FALSE
  )
  return(rbind(system_df, new_entry))
}

# Function: Search System by Topic
search_by_topic <- function(system_df, search_term) {
  results <- system_df[grep(search_term, system_df$Topic, ignore.case = TRUE), ]
  if (nrow(results) == 0) {
    cat(sprintf("No records found matching topic: '%s'\n", search_term))
    return(NULL)
  }
  return(results)
}

# Function: Evidence Strength Visualization
visualize_evidence_strength <- function(system_df) {
  # Set factor levels for structured plot order
  system_df$Strength <- factor(system_df$Strength, levels = c("Weak", "Moderate", "Strong"))

  plot <- ggplot(system_df, aes(x = Topic, fill = Strength)) +
    geom_bar(stat = "count", position = "dodge", color = "black") +
    facet_wrap(~ Position, scales = "free_x") +
    scale_fill_manual(values = c("Strong" = "#2b83ba", "Moderate" = "#fdae61", "Weak" = "#d7191c")) +
    labs(
      title = "Evidence Strength Distribution Across Debate Topics",
      x = "Debate Topic",
      y = "Record Count",
      fill = "Evidence Strength"
    ) +
    theme_minimal() +
    theme(
      axis.text.x = element_text(angle = 45, hjust = 1, size = 9),
      plot.title = element_text(face = "bold", size = 12)
    )

  print(plot)
}

# Function: Formatted Final Debate Report
generate_debate_report <- function(system_df) {
  cat("\n" , rep("=", 70), "\n", sep = "")
  cat("                      FINAL DEBATE SYSTEM REPORT                     \n")
  cat(rep("=", 70), "\n\n", sep = "")

  cat("SUMMARY METRICS:\n")
  cat(sprintf("  • Total Arguments : %d\n", nrow(system_df)))
  cat(sprintf("  • Position 'For'  : %d\n", sum(system_df$Position == "For")))
  cat(sprintf("  • Position 'Against': %d\n", sum(system_df$Position == "Against")))
  cat(sprintf("  • Strong Evidence : %d\n", sum(system_df$Strength == "Strong")))
  cat(sprintf("  • Moderate Evidence: %d\n", sum(system_df$Strength == "Moderate")))
  cat("\n" , rep("-", 70), "\n", sep = "")

  cat("DETAILED DEBATE RECORDS:\n")
  cat(rep("-", 70), "\n\n", sep = "")

  for (i in 1:nrow(system_df)) {
    cat(sprintf("ID: %s | Position: %-7s | Topic: %s\n",
                system_df$Argument_ID[i],
                system_df$Position[i],
                system_df$Topic[i]))
    cat(sprintf("  Claim    : %s\n", system_df$Claim[i]))
    cat(sprintf("  Evidence : %s [%s Type] — Strength: %s\n",
                system_df$Evidence_ID[i],
                system_df$Evidence_Type[i],
                system_df$Strength[i]))
    cat(rep(".", 70), "\n", sep = "")
  }
}


# ==============================================================================
# 3. DEMONSTRATION & EXECUTION
# ==============================================================================

# 1. Search by Topic Demonstration
cat("--- SEARCH RESULT FOR TOPIC: 'Learning' ---\n")
print(search_by_topic(debate_system, "Learning"))

# 2. Render Evidence Strength Plot
visualize_evidence_strength(debate_system)

# 3. Generate Formatted Report
generate_debate_report(debate_system)
```
